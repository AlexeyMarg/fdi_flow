# Пример использования NonlinearityApproximator

В этом примере показано, как аппроксимировать нелинейную динамику временного ряда с помощью многоуровневой линейной регрессии (Ridge/Lasso).

```python
import numpy as np
from fdi_flow.dynamics_approximation.nonlinearity_approximator import MultilevelLinearRegressor
import matplotlib.pyplot as plt
```

```python
# Генерируем синтетические данные
np.random.seed(42)
n_samples = 300
X = np.linspace(-3, 3, n_samples).reshape(-1, 1)
y_true = np.sin(X).ravel()
y = y_true + 0.2 * np.random.randn(n_samples)
```

```python
# Обучаем многоуровневую регрессию
model = MultilevelLinearRegressor(levels=3, model_type='ridge', alpha=0.5)
model.fit(X, y)
y_pred = model.predict(X)
```

```python
# Визуализация
plt.figure(figsize=(8, 4))
plt.plot(X, y, 'o', label='Данные', alpha=0.4)
plt.plot(X, y_true, label='Истинная функция', linewidth=2)
plt.plot(X, y_pred, label='Аппроксимация (MLR)', linewidth=2)
plt.legend()
plt.title('Аппроксимация нелинейности многоуровневой регрессией')
plt.show()
```
